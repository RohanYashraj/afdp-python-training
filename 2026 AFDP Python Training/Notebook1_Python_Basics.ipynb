## 1. Variables: giving values a name

A variable is a label attached to a value, so you can use the value later by name. Python works out the type of the value for you - you do not need to declare it, unlike in C or Java.

```python
policyholder = "Asha Verma"     # a string (text)
age = 45                        # an integer (whole number)
sum_assured = 500000            # another integer
premium_rate = 0.0125           # a float (decimal number)
is_smoker = False               # a boolean (True or False)

print("Policyholder:", policyholder)
print("Age:", age)
print("Sum assured:", sum_assured)
print("Smoker?", is_smoker)

# type() tells you what kind of value a variable holds
print(type(age), type(premium_rate), type(policyholder))
```

**Points to notice**

- Variable names are yours to choose. `sum_assured` is clearer than `s`, and Python does not mind long names.
- Python is *case-sensitive*: `Age` and `age` are two different variables.
- `type()` is handy when something behaves unexpectedly - a "number" read from a file is often actually text.

## 2. Python as a calculator

The usual operators work as you would expect. Two that surprise people: `**` is "to the power of", and `/` always gives a decimal answer even when the numbers divide exactly.

```python
annual_premium = sum_assured * premium_rate
print("Annual premium:", annual_premium)

# Present value of 1 payable in 10 years at 6% p.a.
i = 0.06
v = 1 / (1 + i)          # discount factor
print("v =", v)
print("v^10 =", v ** 10)

# Rounding the answer for presentation
print("v^10 rounded:", round(v ** 10, 4))

# Integer division and remainder are occasionally useful (e.g. converting months to years)
months = 30
print("Years:", months // 12, "Months left over:", months % 12)
```

### A neater way to print: f-strings

Putting an `f` in front of a string lets you drop variables straight into it inside curly brackets. `:.2f` means "show 2 decimal places", and `:,` adds thousand separators. You will use this constantly.

```python
print(f"{policyholder}, aged {age}, pays a premium of {annual_premium:,.2f} per year.")
print(f"The 10-year discount factor at {i:.0%} is {v**10:.4f}.")
```

## 3. Making decisions with `if`, `elif`, `else`

Indentation is not decoration in Python - the indented lines *belong* to the `if`. Four spaces is the convention, and Colab indents for you when you press Enter after the colon.

```python
bmi = 31.5   # change this value and run the cell again

if bmi < 18.5:
    category = "Underweight"
elif bmi < 25:
    category = "Normal"
elif bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"BMI {bmi} -> {category}")
```

```python
# Conditions can be combined with 'and' / 'or'.
# A very simple underwriting rule:
if age > 60 and is_smoker:
    loading = 0.50
elif age > 60 or is_smoker:
    loading = 0.25
else:
    loading = 0.0

print(f"Underwriting loading: {loading:.0%}")
```

## 4. Lists: several values in one variable

A list is an ordered collection in square brackets. Positions start at **0**, not 1 - this catches everyone out at least once.

```python
claims = [12000, 4500, 0, 87000, 2300]

print("All claims:", claims)
print("First claim:", claims[0])
print("Last claim:", claims[-1])
print("Number of claims:", len(claims))
print("Total:", sum(claims))
print("Largest:", max(claims))
print("Average:", sum(claims) / len(claims))

claims.append(15000)      # add a new claim to the end
print("After adding one more:", claims)
```

## 5. Repeating work with loops

A `for` loop runs the indented block once for each item. `range(1, 6)` gives the numbers 1 to 5 - the end value is *not* included.

```python
for year in range(1, 6):
    print(f"Year {year}: discount factor = {v ** year:.4f}")
```

```python
# Looping over a list directly
total = 0
for claim in claims:
    if claim > 0:
        total = total + claim
print(f"Total of non-zero claims: {total:,}")
```

### Actuarial example: building a mortality table with a loop

Below are illustrative one-year mortality rates ($q_x$) for ages 60 to 65. Starting from a radix of 100,000 lives, the loop works out how many survive to each age - exactly the $l_x$ column of a life table.

```python
ages = [60, 61, 62, 63, 64, 65]
qx   = [0.0080, 0.0089, 0.0099, 0.0111, 0.0124, 0.0139]   # illustrative rates

lx = 100000            # radix: lives at age 60
print(f"{'Age':>4} {'qx':>8} {'lx':>10} {'dx':>8}")

for age_x, q in zip(ages, qx):          # zip walks through two lists side by side
    dx = round(lx * q)                   # deaths during the year
    print(f"{age_x:>4} {q:>8.4f} {lx:>10,} {dx:>8,}")
    lx = lx - dx                         # survivors move to the next age

print(f"\nSurvivors at age 66: {lx:,}")
```

`while` loops are the other kind: they keep going *until* a condition stops being true. Useful when you do not know in advance how many steps you need - for example, "how many years until the fund is exhausted?"

```python
fund = 100000
withdrawal = 12000
i = 0.05
year = 0

while fund > 0:
    fund = fund * (1 + i) - withdrawal
    year += 1                     # shorthand for year = year + 1

print(f"The fund is exhausted after {year} years.")
```

## 6. Functions: write once, reuse everywhere

When you find yourself copying the same block of code, turn it into a function. `def` starts the definition, the names in brackets are the *inputs*, and `return` sends the *output* back. The text in triple quotes is a *docstring* - a note explaining what the function does.

```python
def discount_factor(i, n):
    """Present value of 1 payable in n years at interest rate i."""
    return (1 + i) ** (-n)

print(discount_factor(0.06, 10))
print(discount_factor(i=0.04, n=20))     # naming the inputs makes calls easier to read
```

### Actuarial example: expected present value of a term assurance

Using the $q_x$ list from before, the function below calculates the EPV of a benefit of 1 payable at the end of the year of death, for a term of $n$ years:

$$A^{1}_{x:\overline{n}|} = \sum_{t=0}^{n-1} v^{t+1}\; {}_tp_x\; q_{x+t}$$

This is a loop, a list, a function and a bit of arithmetic - everything from this notebook in one place.

```python
def epv_term_assurance(qx_list, i, sum_assured=1):
    """EPV of a term assurance paying sum_assured at the end of the year of death.
    qx_list gives q_x, q_{x+1}, ... for each year of the term."""
    v = 1 / (1 + i)
    tpx = 1.0            # probability of surviving to the start of year t (starts at 1)
    epv = 0.0
    for t, q in enumerate(qx_list):          # enumerate gives you the position t and the value q
        epv += v ** (t + 1) * tpx * q
        tpx *= (1 - q)                        # survive this year -> move to the next
    return epv * sum_assured

epv = epv_term_assurance(qx, i=0.05, sum_assured=500000)
print(f"EPV of a 6-year term assurance for 500,000 at 5%: {epv:,.2f}")
```

```python
# Because it is a function, sensitivity testing is one loop away
for rate in [0.03, 0.04, 0.05, 0.06]:
    print(f"i = {rate:.0%}: EPV = {epv_term_assurance(qx, rate, 500000):,.2f}")
```
