## Defining and Using Functions

Functions are blocks of reusable code that perform a specific task. They help in organizing code, making it more modular and easier to maintain.


```python
# Defining a simple function to calculate the square of a number
def square(num):
    """Returns the square of a given number."""
    return num ** 2

# Using the function
result = square(5)
print("The square of 5 is:", result)
```

### Explanation:
- We define a function using the `def` keyword, followed by the function name and parameters in parentheses.
- The function `square` takes one argument, `num`, and returns its square using the `**` operator.
- Docstrings (`""" ... """`) are used to describe what the function does. This is a good practice, especially when writing more complex functions.
- We then call the function by passing `5` as the argument and print the result.

Functions are fundamental to writing clean and efficient code. They allow you to encapsulate logic and reuse it whenever needed.


## Working with Files

File handling in Python allows you to read from and write to files, enabling you to store and retrieve data efficiently. Let's start by writing data to a file and then reading it back.

```python
# Writing to a file
with open('example.txt', 'w') as file:
    file.write("Hello, world!\n")
    file.write("This is a file handling example.\n")

# Reading from the file
with open('example.txt', 'r') as file:
    content = file.read()

print("File Contents:\n", content)
```

### Explanation:
- We use the `open()` function to open a file. The first argument is the filename, and the second argument is the mode (`'w'` for writing, `'r'` for reading, etc.).
- The `with` statement ensures that the file is properly closed after its suite finishes, even if an exception is raised. This is considered best practice in file handling.
- We wrote two lines of text to `example.txt` using the `write()` method.
- We then read the content of the file using the `read()` method and printed it.

File handling is essential when working with data that needs to be persisted between program runs or shared with other systems.

## Managing Errors with Exception Handling

Errors and exceptions are inevitable in programming. Python provides a way to handle these errors gracefully using `try`, `except`, and `finally` block.


```python
# Example of exception handling
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print("Result:", result)
except ValueError:
    print("That's not a valid number!")
except ZeroDivisionError:
    print("Cannot divide by zero!")
finally:
    print("Execution completed.")
```

### Explanation:
- The `try` block contains code that might raise an exception.
- If an error occurs, the `except` block is executed. Here, we handle two types of exceptions:
  - `ValueError`: Raised if the input is not a valid integer.
  - `ZeroDivisionError`: Raised if the user attempts to divide by zero.
- The `finally` block is optional and contains code that runs no matter what, even if an exception was raised. This is useful for cleaning up resources, like closing files or releasing memory.

### Learning More About Exceptions:
- Python has many built-in exceptions, each designed to handle specific types of errors.
- To learn more about the types of exceptions that can be thrown in Python, you can refer to the official Python documentation. The documentation provides a comprehensive list of exceptions and their use cases: [Python Built-in Exceptions](https://docs.python.org/3/library/exceptions.html).

## Putting It All Together

Let's combine functions, file handling, and exception handling into a single example to reinforce these concepts.


```python
# Function to write a list of numbers to a file and then read and return their squares
def process_numbers(numbers, filename):
    try:
        with open(filename, 'w') as file:
            for num in numbers:
                file.write(f"{num}\n")
        
        with open(filename, 'r') as file:
            squares = [int(line.strip()) ** 2 for line in file]
        
        return squares
    except Exception as e:
        print(f"An error occurred: {e}")
        return []

# Example usage
numbers = [1, 2, 3, 4, 5]
squares = process_numbers(numbers, 'numbers.txt')
print("Squares:", squares)
```

### Explanation:
- We define a function `process_numbers` that takes a list of numbers and a filename as inputs.
- The function writes the numbers to the file and then reads them back, calculating their squares.
- We use exception handling to catch any errors that might occur during file operations.
- Finally, we call the function with a list of numbers and print the squares.

If you want to view the `numbers.txt` file created by this function, you can follow the same process as before:

  - **On Your Computer**: Navigate to the directory where your Jupyter notebook is running using your file explorer (e.g., File Explorer on Windows, Finder on Mac) to locate and open `numbers.txt`.

Viewing the file will help you verify that the numbers were correctly written to the file and that their squares were accurately calculated.
