## Using Keyboard Shortcuts in Jupyter Notebook

To help you work more efficiently in Jupyter Notebook, it's useful to learn some keyboard shortcuts. These shortcuts can save time and help you navigate and edit your notebook more easily.

### Important Considerations:
- **Language and Keyboard Layout**: This course is taught in English, and the shortcuts provided assume a standard English (US) keyboard layout. If you use a different keyboard layout, the locations of some keys may differ. You may need to adapt the shortcuts accordingly.
- **Browser Environment**: Since you’ll be running Jupyter Notebook from your internet browser, make sure that the browser’s shortcuts don't conflict with the ones for Jupyter. If a Jupyter shortcut doesn’t work as expected, check if your browser is intercepting the command.

### Basic Shortcuts:

1. **Switching Between Edit and Command Modes**:
   - **Enter**: Switch to *Edit* mode (allows you to edit a cell).
   - **Esc**: Switch to *Command* mode (allows you to navigate between cells).

2. **Running Code**:
   - **Shift + Enter**: Run the current cell and move to the next cell.
   - **Ctrl + Enter**: Run the current cell and stay in the same cell.

3. **Inserting and Deleting Cells**:
   - **A**: Insert a cell *above* the current cell (in Command mode).
   - **B**: Insert a cell *below* the current cell (in Command mode).
   - **D, D**: Delete the current cell (press 'D' twice quickly in Command mode).

4. **Saving Your Work**:
   - **Ctrl + S** (Cmd + S on Mac): Save the notebook. Note that some browsers automatically save your progress as well.

5. **Copying and Pasting Cells**:
   - **C**: Copy the current cell (in Command mode).
   - **V**: Paste the copied cell *below* the current cell (in Command mode).

6. **Undoing and Redoing**:
   - **Z**: Undo the last action in the notebook (in Command mode).
   - **Y**: Redo the last undone action (in Command mode).

7. **Finding and Replacing Text**:
   - **Ctrl + F** (Cmd + F on Mac): Open the find dialog to search for text in the notebook. Be aware that this might also search the entire browser page, so use the Jupyter-specific dialog if it appears.

### Practice Exercise:
Try using these shortcuts as you navigate through the notebook. For example, use `Shift + Enter` to run a cell, then use `A` or `B` to insert a new cell. Practice switching between *Edit* and *Command* modes with `Enter` and `Esc`.

## Introduction to Python

Before we dive into coding, here are a few key details about Python that will help you as you learn:

- **Dynamically Typed**: Python is a dynamically typed language, which means you don't need to declare the type of a variable when you create it. The type is determined at runtime based on the value assigned to the variable.
  
- **Indentation Matters**: In Python, indentation is not just for readability—it defines the structure of the code. Blocks of code, like those within loops, functions, or conditionals, must be indented consistently. Incorrect indentation will result in errors.

- **Simple Syntax**: Python's syntax is designed to be clear and readable. This simplicity makes it a great choice for beginners and allows you to focus on learning programming concepts rather than dealing with complex syntax.

- **Interpreted Language**: Python is an interpreted language, meaning you write and run the code directly without needing to compile it first. This makes it easy to test and iterate on your code.

Understanding these fundamental aspects of Python will help you as you begin writing your first lines of code.

## Writing Your First Python Code

Python is known for its simple and easy-to-read syntax. Let's start with a classic first program: printing "Hello, World!" to the screen.


```python
# This is a comment. Comments are lines that Python ignores when running the code.
# Comments are useful for explaining what your code does.

# Let's print "Hello, World!" to the screen
print("Hello, World!")
```

### Explanation:
- `print()` is a function that outputs whatever is inside the parentheses to the screen.
- The text inside the quotes `"Hello, World!"` is a string, which is a type of data in Python.

Try changing the text inside the quotes and see what happens when you run the code!


## Variables and Data Types

In Python, variables are used to store information that you can use later in your code. Python is dynamically typed, which means you don't need to declare the type of variable explicitly; Python figures it out for you.


```python
# Let's create some variables
name = "Alice"  # This is a string
age = 25  # This is an integer
height = 5.7  # This is a float

# Now, let's print these variables
print("Name:", name)
print("Age:", age)
print("Height:", height)
```

### Explanation:
- `name` is a variable that holds a string value, `"Alice"`.
- `age` is a variable that holds an integer value, `25`.
- `height` is a variable that holds a floating-point number, `5.7`

As you can see, Python can handle different types of data, such as strings, integers, and floats, and you can store them in variables for later use.

## Basic Arithmetic Operations

Python can also be used as a calculator. You can perform arithmetic operations like addition, subtraction, multiplication, and division.


```python
# Basic arithmetic operations
addition = 10 + 5  # Adds 10 and 5
subtraction = 10 - 5  # Subtracts 5 from 10
multiplication = 10 * 5  # Multiplies 10 by 5
division = 10 / 5  # Divides 10 by 5

# Let's print the results
print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

### Explanation:
- The `+`, `-`, `*`, and `/` symbols are used for addition, subtraction, multiplication, and division, respectively.
- Python will automatically compute the results and you can store them in variables.

Experiment with other numbers and operations to see how Python handles them.

## Control Structures: Conditional Statements

Control structures allow you to control the flow of your code. One of the most basic control structures is the conditional statement, which lets you execute code only if certain conditions are met.


```python
# Let's check if a number is positive, negative, or zero
number = 10  # You can change this value and see what happens

if number > 0:
    print("The number is positive.")
elif number == 0:
    print("The number is zero.")
else:
    print("The number is negative.")
```

### Explanation:
- `if` checks if a condition is true; if it is, the code indented underneath it runs.
- `elif` (short for "else if") checks another condition if the first `if` is false.
- `else` runs if none of the previous conditions are true.

Try changing the value of `number` to see how the output changes.


## Control Structures: Loops

Loops allow you to repeat code multiple times. The most common types of loops are `for` loops and `while` loops.


```python
# Using a for loop to print numbers 1 to 5
for i in range(1, 6):
    print(i)
```

### Explanation:
- `for` loops allow you to iterate over a sequence (like a list or range of numbers).
- `range(1, 6)` generates numbers from 1 to 5 (6 is not included).

This loop prints the numbers 1 through 5.


```python
# Using a while loop to print numbers 1 to 5
i = 1  # Start with 1
while i <= 5:
    print(i)
    i += 1  # Increase i by 1 each time
```

### Explanation:
- `while` loops continue to execute as long as a condition is true.
- The loop above keeps running until `i` is greater than 5.

You can use loops to automate repetitive tasks in your code.
